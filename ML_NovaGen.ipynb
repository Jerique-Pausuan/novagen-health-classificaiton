```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import recall_score, classification_report, accuracy_score
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
df = pd.read_csv("novagen_dataset.csv")
df
```

```python
x = df.drop(columns=["Target"])
y = df["Target"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
x_train_scale = scaler.fit_transform(x_train)
x_test_scale = scaler.transform(x_test)
```

BASELINE MODEL

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(
    penalty="l2",
    max_iter=1000,
    solver="liblinear"
)
lr.fit(x_train_scale, y_train)
y_pred = lr.predict(x_test_scale)
print("recall score:", recall_score(y_test, y_pred))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("classification report:", classification_report(y_test, y_pred))
```

MODEL 2

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(
    n_estimators=200,
    max_depth=3,
    random_state=42
)
rfc.fit(x_train_scale, y_train)
y_pred = rfc.predict(x_test_scale)
print("recall score:", recall_score(y_test, y_pred))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("classification report:", classification_report(y_test, y_pred))
```

MODEL 3

```python
 from sklearn.svm import SVC
svc = SVC(
    kernel='rbf',
    C=5
)
svc.fit(x_train_scale, y_train)
y_pred = svc.predict(x_test_scale)
print("recall score:", recall_score(y_test, y_pred))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("classification report:", classification_report(y_test, y_pred))
```

MODEL 4

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)
knn.fit(x_train_scale, y_train)
y_pred = knn.predict(x_test_scale)
print("recall score:", recall_score(y_test, y_pred))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("classification report:", classification_report(y_test, y_pred))
```

GRADIENT BOOSTING

```python
from sklearn.ensemble import GradientBoostingClassifier
gbc = GradientBoostingClassifier(
    n_estimators=150,
    random_state=42,
    learning_rate=0.1,
    max_depth=3
)
gbc.fit(x_train_scale, y_train)
y_pred = gbc.predict(x_test_scale)
print("recall score:", recall_score(y_test, y_pred))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("classification report:", classification_report(y_test, y_pred))
```

VOTING CLASSFIER

```python
from sklearn.ensemble import VotingClassifier
lr = LogisticRegression(
    penalty="l2",
    max_iter=1000,
    solver="liblinear"
)

rfc = RandomForestClassifier(
    n_estimators=200,
    max_depth=3,
    random_state=42
)

svc = SVC(
    kernel='rbf',
    C=1
)

knn = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

vc = VotingClassifier(estimators=[
    ("lr", lr),
    ("rfc", rfc),
    ("svc", svc),
    ("knn", knn)
])

vc.fit(x_train_scale, y_train)
y_pred = vc.predict(x_test_scale)
print("recall score:", recall_score(y_test, y_pred))
print("accuracy score:", accuracy_score(y_test, y_pred))
print("classification report:", classification_report(y_test, y_pred))
```
